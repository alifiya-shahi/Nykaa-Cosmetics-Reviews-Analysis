```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import seaborn as sns

sns.set_theme(style="darkgrid")

print("Libraries imported successfully!")
```

```python
data = pd.read_csv("/content/nyka_top_brands_cosmetics_product_reviews.csv", on_bad_lines='skip', engine='python')

print("Dataset loaded successfully!")
```

```python
print("Shape of dataset:", data.shape)

print("\nColumns:")
print(data.columns.tolist())

print("\nDataset Information:")
data.info()
```

```python
data.tail()
```

```python
data.describe()
```

```python
missing_values = data.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
missing_percentage = (data.isnull().sum() / len(data)) * 100

print("Missing value percentage:")
print(missing_percentage)
```

```python
duplicate_values = data.duplicated().sum()

print("Number of duplicate rows:", duplicate_values)
```

```python
data["review_date"] = pd.to_datetime(data["review_date"])

print("Earliest review:", data["review_date"].min())
print("Latest review:", data["review_date"].max())
```

```python
# Complete initial data check

print("===== DATASET SHAPE =====")
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])

print("\n===== COLUMN NAMES =====")
print(data.columns.tolist())

print("\n===== DATA TYPES =====")
print(data.dtypes)

print("\n===== MISSING VALUES =====")
print(data.isnull().sum())

print("\n===== DUPLICATE ROWS =====")
print(data.duplicated().sum())

print("\n===== UNIQUE VALUES =====")
print(data.nunique())

print("\n===== STATISTICAL SUMMARY =====")
display(data.describe(include="all").T)
```

```python
# ==========================================
# DATA QUALITY CHECK
# ==========================================

print("===== MISSING VALUE PERCENTAGE =====")

missing = data.isnull().sum()
missing_percent = (missing / len(data)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percent.round(2)
})

display(missing_table)


print("\n===== CHECK PRICE VS MRP =====")

price_problem = data[data["price"] > data["mrp"]]

print("Rows where Price > MRP:", len(price_problem))

display(
    price_problem[
        ["brand_name", "product_title", "mrp", "price"]
    ].head(20)
)


print("\n===== CHECK REVIEW LABEL =====")

print("Review Label:")
print(data["review_label"].value_counts(dropna=False))


print("\n===== REVIEW LABEL vs BUYER STATUS =====")

display(
    pd.crosstab(
        data["is_a_buyer"],
        data["review_label"],
        dropna=False
    )
)


print("\n===== CHECK PRODUCT TAGS =====")

print("Available product tags:",
      data["product_tags"].notna().sum())

print("Missing product tags:",
      data["product_tags"].isna().sum())
```

```python
# ==========================================
# DATA CLEANING
# ==========================================

# Make a copy of the original dataset
clean_data = data.copy()

# ------------------------------------------
# 1. Handle missing review text
# ------------------------------------------

clean_data["review_text"] = clean_data["review_text"].fillna("No Review Text")


# ------------------------------------------
# 2. Handle missing review labels
# ------------------------------------------

clean_data["review_label"] = clean_data["review_label"].fillna("Not Available")


# ------------------------------------------
# 3. Remove product_tags
# ------------------------------------------

clean_data = clean_data.drop(columns=["product_tags"])


# ------------------------------------------
# 4. Convert review_date to datetime
# ------------------------------------------

clean_data["review_date"] = pd.to_datetime(
    clean_data["review_date"],
    errors="coerce"
)


# ------------------------------------------
# 5. Check duplicates
# ------------------------------------------

clean_data = clean_data.drop_duplicates()


# ==========================================
# CLEANED DATASET CHECK
# ==========================================

print("===== CLEANED DATASET ====")

print("Rows:", clean_data.shape[0])
print("Columns:", clean_data.shape[1])

print("\n===== MISSING VALUES ====")
print(clean_data.isnull().sum())

print("\n===== DUPLICATE ROWS ====")
print(clean_data.duplicated().sum())

print("\n===== DATA TYPES ====")
print(clean_data.dtypes)

print("\n===== FIRST 5 ROWS ====")
display(clean_data.head())
```

```python
# ==========================================
# BASIC DATA ANALYSIS
# ==========================================

print("===== PROJECT SUMMARY =====")
print("Total Reviews:", len(clean_data))
print("Total Products:", clean_data["product_id"].nunique())
print("Total Brands:", clean_data["brand_name"].nunique())

print("\n===== BRAND-WISE REVIEWS =====")
print(clean_data["brand_name"].value_counts())

print("\n===== RATING DISTRIBUTION =====")
print(clean_data["review_rating"].value_counts().sort_index())

print("\n===== BUYER STATUS =====")
print(clean_data["is_a_buyer"].value_counts())

print("\n===== AVERAGE VALUES =====")
print("Average Review Rating:",
      round(clean_data["review_rating"].mean(), 2))

print("Average Product Rating:",
      round(clean_data["product_rating"].mean(), 2))

print("Average MRP:",
      round(clean_data["mrp"].mean(), 2))

print("Average Price:",
      round(clean_data["price"].mean(), 2))
```

```python
# ==========================================
# DATA VISUALIZATION
# ==========================================

# 1. Reviews by Brand
brand_counts = clean_data["brand_name"].value_counts()

plt.figure(figsize=(10, 5))
brand_counts.plot(kind="bar")
plt.title("Number of Reviews by Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 2. Review Rating Distribution
rating_counts = clean_data["review_rating"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
rating_counts.plot(kind="bar")
plt.title("Review Rating Distribution")
plt.xlabel("Review Rating")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# 3. MRP vs Selling Price
price_data = clean_data[["mrp", "price"]].head(20)

plt.figure(figsize=(10, 5))
plt.plot(price_data["mrp"].values, marker="o", label="MRP")
plt.plot(price_data["price"].values, marker="o", label="Selling Price")
plt.title("MRP vs Selling Price (First 20 Products)")
plt.xlabel("Product")
plt.ylabel("Price (₹)")
plt.legend()
plt.tight_layout()
plt.show()


# 4. Buyer vs Non-Buyer
buyer_counts = clean_data["is_a_buyer"].value_counts()

plt.figure(figsize=(7, 5))
buyer_counts.plot(kind="bar")
plt.title("Buyer vs Non-Buyer Reviews")
plt.xlabel("Is a Buyer")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# 5. Average Product Rating by Brand
brand_rating = clean_data.groupby("brand_name")["product_rating"].mean().sort_values()

plt.figure(figsize=(10, 5))
brand_rating.plot(kind="bar")
plt.title("Average Product Rating by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Product Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# 1. WHICH BRANDS HAVE THE MOST REVIEWS?

```python
plt.figure(figsize=(10, 5))

brand_reviews = clean_data["brand_name"].value_counts().sort_values(ascending=False)

sns.barplot(x=brand_reviews.index, y=brand_reviews.values)

plt.title("Number of Reviews by Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# 2. ARE CUSTOMERS MOSTLY GIVING 5-STAR RATINGS?


```python
plt.figure(figsize=(8, 5))

rating_counts = clean_data["review_rating"].value_counts().sort_index()

sns.barplot(x=rating_counts.index, y=rating_counts.values)

plt.title("Customer Review Rating Distribution")
plt.xlabel("Review Rating")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()
```

# 3. WHAT IS THE TYPICAL PRODUCT PRICE?


```python
plt.figure(figsize=(9, 5))
plt.hist(clean_data["price"], bins=30)
plt.title("Product Price Distribution")
plt.xlabel("Price (₹)")
plt.ylabel("Number of Products/Reviews")
plt.tight_layout()
plt.show()
```

# 4. HOW MANY REVIEWERS ARE ACTUAL BUYERS?

```python
plt.figure(figsize=(7, 5))

buyer_counts = clean_data["is_a_buyer"].value_counts()

sns.barplot(x=buyer_counts.index, y=buyer_counts.values)

plt.title("Buyer vs Non-Buyer")
plt.xlabel("Is a Buyer")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()
```

# 5. HOW MANY ARE PRO USERS?

```python
pro_counts = clean_data["pro_user"].value_counts()

plt.figure(figsize=(7, 5))
pro_counts.plot(kind="bar")
plt.title("Pro User vs Non-Pro User")
plt.xlabel("Pro User")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# 6. VERIFIED BUYER vs VERIFIED PROFESSIONAL vs NOT AVAILABLE

```python
plt.figure(figsize=(9, 5))

label_counts = clean_data["review_label"].value_counts()

sns.barplot(x=label_counts.index, y=label_counts.values)

plt.title("Review Label Distribution")
plt.xlabel("Review Label")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

# 7. WHAT PERIOD DO THE REVIEWS COVER?

```python
monthly_reviews = (
    clean_data
    .set_index("review_date")
    .resample("ME")
    .size()
)

plt.figure(figsize=(12, 5))
monthly_reviews.plot()
plt.title("Number of Reviews Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# FINAL KEY NUMBERS
# ==========================================

print("===== KEY FINDINGS =====")

print("\nTop Brand:")
print(clean_data["brand_name"].value_counts().idxmax())

print("\nTop Brand Review Count:")
print(clean_data["brand_name"].value_counts().max())

print("\n5-Star Reviews:")
print((clean_data["review_rating"] == 5).sum())

print("\n5-Star Review Percentage:")
print(round((clean_data["review_rating"] == 5).mean() * 100, 2), "%")

print("\nAverage Price:")
print("₹", round(clean_data["price"].mean(), 2))

print("\nMedian Price:")
print("₹", round(clean_data["price"].median(), 2))

print("\nActual Buyers:")
print(clean_data["is_a_buyer"].sum())

print("\nNon-Buyers:")
print((clean_data["is_a_buyer"] == False).sum())

print("\nPro Users:")
print(clean_data["pro_user"].sum())

print("\nNon-Pro Users:")
print((clean_data["pro_user"] == False).sum())

print("\nReview Labels:")
print(clean_data["review_label"].value_counts())

print("\nReview Period:")
print(clean_data["review_date"].min())
print("to")
print(clean_data["review_date"].max())
```
